## 1. Unzip and fix `data.yaml` paths
Unzips to Colab's local disk (faster than Drive) and rewrites the paths so training works from anywhere.

```python
import zipfile, yaml

DATA = Path('/content/datasets')
def prepare(key):
    dest = DATA / key
    if not dest.exists():
        dest.mkdir(parents=True)
        with zipfile.ZipFile(ZIPS[key]) as zf:
            zf.extractall(dest)
    yamls = list(dest.rglob('data.yaml'))
    assert yamls, f'no data.yaml found for {key}'
    root = yamls[0].parent
    d = yaml.safe_load((root / 'data.yaml').read_text())
    d['path'] = str(root)
    for k, sub in (('train', 'train'), ('val', 'valid'), ('test', 'test')):
        if (root / sub / 'images').exists():
            d[k] = f'{sub}/images'
        else:
            d.pop(k, None)
    (root / 'data.yaml').write_text(yaml.safe_dump(d, sort_keys=False))
    n = {s: len(list((root / s / 'images').glob('*'))) for s in ('train', 'valid', 'test') if (root / s / 'images').exists()}
    print(f'{key}: {len(d["names"])} classes, images per split: {n}')
    print('  classes:', d['names'])
    return root / 'data.yaml'

PARKING_YAML = prepare('parking')
```

## 2b. Re-split for v3
Validation = ONLY our own clips, so progress on our footage decides which weights are kept. The original dataset's validation images move into train. Our own training crops are repeated 3x so they count more.

```python
import shutil, glob, os
root = PARKING_YAML.parent
moved = 0
for img in glob.glob(f'{root}/valid/images/*'):
    if os.path.basename(img).startswith('my_'):
        continue
    base = os.path.splitext(os.path.basename(img))[0]
    lab = f'{root}/valid/labels/{base}.txt'
    shutil.move(img, f'{root}/train/images/{os.path.basename(img)}')
    if os.path.exists(lab): shutil.move(lab, f'{root}/train/labels/{base}.txt')
    moved += 1
dup = 0
for img in glob.glob(f'{root}/train/images/my_*.jpg'):
    base = os.path.splitext(os.path.basename(img))[0]
    if '_rep' in base: continue
    lab = f'{root}/train/labels/{base}.txt'
    for k in (2, 3):
        shutil.copy(img, f'{root}/train/images/{base}_rep{k}.jpg')
        if os.path.exists(lab): shutil.copy(lab, f'{root}/train/labels/{base}_rep{k}.txt')
        dup += 1
print('moved to train:', moved, '| extra copies of our crops:', dup)
for s in ('train', 'valid', 'test'):
    print(s, len(os.listdir(f'{root}/{s}/images')), 'images', len(os.listdir(f'{root}/{s}/labels')), 'labels')
```

## 2. Training helper

```python
from ultralytics import YOLO

def train(data_yaml, name, base='yolo26n.pt', imgsz=640, epochs=40, batch=16):
    model = YOLO(base)  # COCO-pretrained, then fine-tuned on your data
    model.train(
        data=str(data_yaml), epochs=epochs, imgsz=imgsz, batch=batch,
        patience=40,                 # effectively no early stop: validation is now OUR clips only
        optimizer='AdamW', lr0=0.0003, lrf=0.1, cos_lr=True,   # gentle fine-tune (v2's auto lr was too high)
        project=str(RUNS), name=name, exist_ok=True, device=0,
    )
    return RUNS / name / 'weights' / 'best.pt'
```

## 3. Train the parking model

```python
parking_best = train(PARKING_YAML, 'parking_v3', base=str(START_WEIGHTS))
print(parking_best)
```

## 5. Evaluate on the held-out **test** split
Note: the parking dataset's own test split likely shares scenes with train (augmented copies), so treat this as an *in-distribution* score. Your real test is your own unseen mall clips, run locally later.

```python
m = YOLO(str(parking_best))
r = m.val(data=str(PARKING_YAML), split='test', device=0)
print(f'== parking_v3: mAP50={r.box.map50:.3f}  mAP50-95={r.box.map:.3f}  P={r.box.mp:.3f}  R={r.box.mr:.3f}')
```
